#### Задача 2. 

Представьте, что проводится голосование за лучшее блюдо в школе. У вас есть результаты голосования, представленные в виде списка, где каждый элемент — это имя ученика и его голос (блюдо).

Пример данных:

    votes = [
        ("Анна", "Пицца"),
        ("Олег", "Бургер"),
        ("Мария", "Пицца"),
        ("Алексей", "Паста"),
        ("Ольга", "Бургер"),
        ("Иван", "Пицца"),
        ("Вика", "Паста")
    ]
    
Задание:

1. Создайте функцию count_votes(votes), которая принимает список голосов и возвращает словарь, где ключом является название блюда, а значением — количество голосов за него.
2. Добавьте в функцию возможность находить блюдо-победитель — то, которое набрало наибольшее количество голосов.
3. Если несколько блюд набрали одинаковое количество голосов, выведите их все.

Пример вывода:

    {
        "Пицца": 3,
        "Бургер": 2,
        "Паста": 2
    }
    Блюда-победители: Пицца

```python
from collections import defaultdict

def count_votes(v):
    vote_counts = defaultdict(int)          # словарь для подсчета голосов
        
    for _, dish in v:                       # игнорируем имя с помощью _
        vote_counts[dish] += 1

    max_votes = max(vote_counts.values())

    winners = [dish for dish, count in vote_counts.items() if count == max_votes]
    # генератор собирает список блюд-победителей 

    print(dict(vote_counts))
    print("блюда-победители:", ", ".join(winners))

votes = [
    ("Анна", "Пицца"),
    ("Олег", "Бургер"),
    ("Мария", "Пицца"),
    ("Алексей", "Паста"),
    ("Ольга", "Бургер"),
    ("Иван", "Пицца"),
    ("Вика", "Паста")
]

count_votes(votes)
```

#### Задача 3.

Вы разрабатываете простой переводчик для учеников младших классов. Вам нужно создать программу, которая будет переводить слова с английского на русский язык с использованием словаря.

Задание:
1. Создайте словарь, в котором ключами будут английские слова, а значениями — их переводы на русский.
   
   Пример словаря:
  
       translations = {
           "cat": "кот",
           "dog": "собака",
           "apple": "яблоко",
           "book": "книга",
           "sun": "солнце"
       }
   
2. Напишите функцию translate_word(word, dictionary), которая принимает английское слово и словарь переводов, и возвращает перевод этого слова на русский. Если слово не найдено, возвращайте сообщение: "Перевод не найден".
   
3. Дополнительно: сделайте программу интерактивной, чтобы пользователь мог вводить английские слова и получать переводы до тех пор, пока он не введет команду "exit".

Пример работы программы:

    Введите слово для перевода: cat
    Перевод: кот

    Введите слово для перевода: banana
    Перевод не найден

    Введите слово для перевода: exit
    Программа завершена.
    
Добавьте возможность расширять словарь, предлагая пользователю вводить новые слова и их переводы, если слово не было найдено.

```python
translations = {
    "cat": "кот",
    "dog": "собака",
    "apple": "яблоко",
    "book": "книга",
    "sun": "солнце"
}

def translate_word(word, dictionary):
    """функция для перевода слова с английского на русский."""
    return dictionary.get(word, "перевод не найден")

def add_translation(word, translation, dictionary):
    """добавляет новое слово и его перевод в словарь."""
    dictionary[word] = translation
    print(f"слово '{word}' добавлено в словарь с переводом '{translation}'")

def main():
    print("вэлком!")
    print("введите 'exit' для выхода")
    
    while True:
        word = input("введите слово для перевода: ").strip().lower()
        
        if word == "exit":
            print("пока!")
            break
        
        translation = translate_word(word, translations)
        if translation == "перевод не найден":
            print(translation)
            add_new = input("хотите добавить это слово? (да/нет): ").strip().lower()
            if add_new == "да":
                new_translation = input("введите перевод для слова '{}': ".format(word)).strip().lower()
                add_translation(word, new_translation, translations)
        else:
            print(f"перевод: {translation}")

main()
```

Обнаружилось, что имена некоторых владельцев повторяются, потому что у них несколько кошек. Необходимо оптимизировать хранение данных таким образом, чтобы для каждого владельца при выводе на печать данные о всех его животных отображались в одной строке:

Игорь Бероев: Муся, 7; Изольда, 2

```python
cats = [('Мартин', 5, 'Алексей', 'Егоров'),
            ('Фродо', 3, 'Анна', 'Самохина'),
            ('Вася', 4, 'Андрей', 'Белов'),
            ('Муся', 7, 'Игорь', 'Бероев'),
            ('Изольда', 2, 'Игорь', 'Бероев'),
            ('Снейп', 1, 'Марина', 'Апраксина'),
            ('Лютик', 4, 'Виталий', 'Соломин'),
            ('Снежок', 3, 'Марина', 'Апраксина'),
            ('Марта', 5, 'Сергей', 'Колесников'),
            ('Буся', 12, 'Алена', 'Федорова'),
            ('Джонни', 10, 'Игорь', 'Андропов'),
            ('Мурзик', 1,'Даниил', 'Невзоров'),
            ('Барсик', 2, 'Виталий', 'Соломин'),
            ('Рыжик', 7, 'Владимир', 'Медведев'),
            ('Матильда', 8, 'Андрей', 'Белов'),
            ('Гарфилд', 3, 'Александр', 'Березуев')]

owners = {}

for cat in cats:
    p_name, p_age, o_name, o_surname = cat
    owner = (o_name, o_surname)                                            # ключ для словаря owners
    if owner not in owners:                                                # если владельца еще нет, добавляем пустой список
        owners[owner] = []
    owners[owner].append((p_name, p_age))                                  # добавляем в список для владельца кортеж с инфой о кошке

for owner, pets in owners.items():                                         # для кортежа owner и списка кортежей pets в словаре
    o_name = f"{owner[0]} {owner[1]}"
    pets_info = '; '.join([f'{name}, {age}' for name, age in pets])        
    print(f'{o_name}: {pets_info}')
```
