#### Задача 1. 

Перепишите один из наших предыдущих классов (лингвиста, компьютерного лингвиста или программиста) таким образом, чтобы у него были два атрибута, управляемых дескриптором: зарплата и количество рабочих часов в неделю. Дескриптор должен не давать устанавливать зарплату меньше какого-то значения (можно поругаться), а количество рабочих часов - больше положенного. 

```python
class Puppet:
    def __init__(self, min_value=None, max_value=None): 
        self.min_value = min_value
        self.max_value = max_value
        self.private_name = None

    def __set_name__(self, owner, name):
        self.private_name = "_" + name

    def __get__(self, instance, owner):
        if instance is None:
            return self
        return getattr(instance, self.private_name)

    def __set__(self, instance, value):
        if self.min_value is not None and value < self.min_value:
            print(f"нельзя установить значение меньше {self.min_value}")
            return
        if self.max_value is not None and value > self.max_value:
            print(f"нельзя установить значение больше {self.max_value}")
            return
        setattr(instance, self.private_name, value)


class Employee:
    """общий класс-родитель"""
    salary = Puppet(min_value=200)           # минимальная зарплата
    hours_per_week = Puppet(max_value=40)    # максимум часов в неделю

    def __init__(self, name, surname, salary=350, hours=30):
        self.name = name
        self.surname = surname

        self.salary = salary            
        self.hours_per_week = hours     

    def work(self, *args):
        raise NotImplementedError


class Linguist(Employee):
    """класс ливсгниста"""
    def __init__(self, name, surname, salary=500, hours=25):
        super().__init__(name, surname, salary, hours)
        self.publications = []

    def work(self, hours, name_of_publ):
        print("пишет...")
        self.publications.append(name_of_publ)
        print(f"{self.name} {self.surname} опубликовал статью {name_of_publ}")
```

```python
n = Linguist("Noam", "Chomsky")
```

```python
n.salary = 100      
n.hours_per_week = 50 
```

```python
n.hours_per_week = 20 
n.salary = 5000  
```

```python
Linguist.work(n, '40', 'Generative something something')
```
