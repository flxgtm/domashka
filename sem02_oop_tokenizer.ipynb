```python
import re
```

```python
class Token:
    def __init__(self, text: str, category: str):
        self.text = text
        self.category = category

    def __repr__(self):
        return f"Token(text='{self.text}', category='{self.category}')"
```

```python
class Tokenizer:
    def __init__(self):
        self.pattern = re.compile(
            r"(?P<latin>[A-Za-z]+)|(?P<cyrillic>[А-Яа-яЁё]+)|(?P<punct>[^\w\s])"
        )

    def tokenize(self, text: str):
        """генератор, возвращающий объекты token"""
        for match in re.finditer(self.pattern, text):
            if match.group("latin"):
                category = "latin"
            elif match.group("cyrillic"):
                category = "cyrillic"
            else:
                category = "punctuation"
            yield Token(match.group(0), category)
```

```python
text = "Создайте класс Artifact, который представляет артефакт. Артефакт должен иметь следующие свойства:"
tokenizer = Tokenizer()

for token in tokenizer.tokenize(text):
    print(token)
```
