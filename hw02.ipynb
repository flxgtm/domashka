## Задание 1.

Вводится список чисел через пробел. Поменяйте в этом списке самое маленькое и самое большое число местами, н-р, список 4 2 6 3 7 5 должен превратиться в 4 7 6 3 2 5

```python
spisok = [int(s) for s in input().split()]
index_of_min = 0
index_of_max = 0
for i in range(1, len(spisok)):
    if spisok[i] > spisok[index_of_max]:
        index_of_max = i
    if spisok[i] < spisok[index_of_min]:
        index_of_min = i
spisok[index_of_min], spisok[index_of_max] = spisok[index_of_max], spisok[index_of_min]
print(' '.join([str(i) for i in spisok]))
```

## Задание 2. 

Вводится список (чего угодно, как угодно). Нужно все элементы поменять местами, н-р, список вида 1 2 3 4 5 превращается в 2 1 4 3 5. 

```python
#я не очень поняла, норм, что тройка на своем месте остается?

def swap_all(lst):
    n = len(lst)
    for i in range(n // 2):
        lst[i], lst[n - 1 - i] = lst[n - 1 - i], lst[i]
    return lst

lst = [1, 2, 3, 4, 5, 6]
swap_all(lst)
print(lst) 
```

## Задание 3. 

Петя перешёл в другую школу. На уроке физкультуры ему понадобилось определить своё место в строю. Помогите ему это сделать.

Программа получает на вход невозрастающую последовательность натуральных чисел, означающих рост каждого человека в строю. После этого вводится число X – рост Пети. Все числа во входных данных натуральные и не превышают 200.

Выведите номер, под которым Петя должен встать в строй. Если в строю есть люди с одинаковым ростом, таким же, как у Пети, то он должен встать после них.

```python
def petya_position(p, h):
    return next((p.index(i)+1 for i in p if i < h), len(p) + 1)
 
pupils = map(int, input().split())
height = int(input())
print(petya_position(sorted(pupils, reverse=True), height))
```

## Задание 4.

Известно, что на доске 8х8 можно расставить 8 ферзей так, чтобы они не били друг друга. Вам дана расстановка 8 ферзей на доске, определите, есть ли среди них пара бьющих друг друга.
Программа получает на вход восемь пар чисел, каждое число от 1 до 8 — координаты 8 ферзей. Если ферзи не бьют друг друга, выведите слово NO, иначе выведите YES.

```python
rows = []
columns = []
for _ in range(8):
    r, c = [int(s) for s in input().split()]
    rows.append(r)
    columns.append(c)
count = 0
for i in range(8):
    for j in range(i+1, 8):
        if rows[i] == rows[j] or columns[i]==columns[j] or abs(rows[i]-rows[j])==abs(columns[i]-columns[j]):
            count += 1
if count==0:
    print("NO")
else:
    print("YES")
    
```

## Задание 5.

Дан список. Удалите из списка все повторы, то есть, оставьте в списке каждый элемент только в одном экземпляре (остаться должно только первое появление элемента в списке). Программа не должна использовать дополнительную память: это означает, что нельзя пользоваться множествами и генераторами, можно только производить изменения в самом списке. 

```python
def remove_duplicates(lst):
    i = 0
    while i < len(lst):
        j = i + 1
        while j < len(lst):
            if lst[i] == lst[j]:
                del lst[j]  
            else:
                j += 1
        i += 1

lst = [1, 2, 7, 4, 1, 2, 8, 3]
remove_duplicates(lst)
print(lst) 
```

## Задание 6.

На вход программе даётся число N и список. Удалите из списка все элементы, которые встречаются в нём ровно N раз. Программа не должна использовать дополнительную память.

```python
def remove_n_occurrences(lst, n):
    for i in range(len(lst) - 1, -1, -1): 
        if lst.count(lst[i]) == n:
            del lst[i]
    return lst

lst = [1, 2, 3, 1, 2, 4, 5, 3, 3]
n = 1
print(remove_n_occurrences(lst, n)) 
```

## Задание 7.

Напишите функцию, которая принимает неограниченное количество аргументов-строк и параметр = метод строки, который затем последовательно применяется к каждой строчке. Результаты можно вернуть списком. Оформите функцию грамотно, напишите docstring. 

```python
def fonksiyon(method, *args):
    """
    Применяет указанный метод строки ко всем переданным строкам.

    :param method: Метод строки, который будет применен.
    :param args: Неограниченное количество строк.
    :return: Список результатов применения метода к каждой строке.
    """
    results = []
    for arg in args:
        results.append(getattr(arg, method)())
    return results

print(fonksiyon('upper', 'hello', 'world'))  # Результат: ['HELLO', 'WORLD']
print(fonksiyon('strip', '   test   ', '   example   '))  # Результат: ['test', 'example']
```

## Задание 8. 

Отсортируйте список строчек в зависимости от того, сколько в каждой строке содержится букв а. (н-р, строчки "мама", "арбуз" и "абракадабра" должны выстроиться в "арбуз", "мама", "абракадабра").

```python
def sort_by_a_count(strings):
    return sorted(strings, key=lambda x: x.count('а'))

strings = ["мама", "арбуз", "абракадабра"]
sorted_strings = sort_by_a_count(strings)
print(sorted_strings)  
```

## Задание 9. 

Напишите функцию-генератор, которая будет возвращать новые функции, возводящие в степень N. Т.е., функция-генератор принимает в качестве аргумента число N, а возвращает функцию, которая будет принимать число и возвращать его же в степени N. 

```python
def power_generator(n):
    def power(x):
        return x ** n
    return power

square = power_generator(2)
cube = power_generator(3)

print(square(5)) 
print(cube(5))    
```

## Задание 10. 

Напишите рекурсивную функцию, которая будет возвращать, является ли строка палиндромом. На вход функция получает строки типа "абба, "абсба" и подобные (пробелов внутри этих строк нет). 

```python
a = str(input())
def palindrom(s):
    if len(s) < 1:
        return True
    if s[0] != s[-1]:
        return False
    return palindrom(s[1:-1])

if (palindrom(a) == True):
    print ('ok')
else:
    print ('no')
```
